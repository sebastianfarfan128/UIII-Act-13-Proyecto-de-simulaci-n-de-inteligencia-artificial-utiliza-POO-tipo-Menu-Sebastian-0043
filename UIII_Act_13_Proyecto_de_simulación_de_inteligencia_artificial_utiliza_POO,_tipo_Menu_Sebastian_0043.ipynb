### Proyecto de simulación de inteligencia artificial  utiliza POO, tipo Menu
### 18. Interés compuesto

```python
# --- DEFINICIÓN DE LA CLASE ---
class Interes_compuesto:
    """Clase que representa una inversión financiera analizada por IA."""

    # 1. Función Constructor (Atributos) [5, 6]
    def __init__(self, cliente, capital, tasa, tiempo):
        # 5 Atributos del objeto [7, 8]
        self.cliente = cliente
        self.capital = float(capital)
        self.tasa = float(tasa)
        self.tiempo = int(tiempo)
        self.monto_final = 0.0 # Se calculará con un método

    # 2. Función de Cálculo Matemático [9, 10]
    def calcular_monto(self):
        """Calcula el monto final usando la fórmula de interés compuesto."""
        # Fórmula: Monto = Capital * (1 + tasa/100) ^ tiempo
        self.monto_final = self.capital * (1 + self.tasa / 100) ** self.tiempo

    # 3. Función de Análisis (Simulación de IA) [11, 12]
    def analizar_inversion_ia(self):
        """La IA evalúa la rentabilidad del proyecto."""
        ganancia = self.monto_final - self.capital
        if ganancia > (self.capital * 0.5):
            return "IA: RENTABILIDAD ALTA (Excelente)"
        elif ganancia > 0:
            return "IA: RENTABILIDAD ESTABLE (Segura)"
        else:
            return "IA: SIN GANANCIAS (Riesgo)"

    # 4. Función de Formateo [3]
    def obtener_fila_tabla(self):
        """Devuelve los datos formateados para una tabla alineada."""
        ganancia = self.monto_final - self.capital
        return f"{self.cliente:<15} | {self.capital:<10.2f} | {self.tasa:<8.1f} | {self.tiempo:<8} | {self.monto_final:<12.2f} | {ganancia:<10.2f}"

    # 5. Función de Resumen Individual [13]
    def mostrar_informe(self):
        """Muestra un reporte detallado del objeto."""
        print(f"\n>>> REPORTE INDIVIDUAL: {self.cliente}")
        print(f"Estado de la IA: {self.analizar_inversion_ia()}")

# --- INICIALIZACIÓN DEL SISTEMA ---
lista_inversiones = [] # La lista comienza vacía

# --- BUCLE PRINCIPAL (MENÚ INTERACTIVO) [14, 15] ---
print("====================================================")
print("     NÚCLEO DE INTELIGENCIA ARTIFICIAL FINANCIERA   ")
print("====================================================")

while True:
    # Lógica de Percepción de la IA
    cantidad = len(lista_inversiones)
    if cantidad == 0:
        estado_ia = "MODO PASIVO: Esperando datos financieros."
    else:
        estado_ia = f"SISTEMA ACTIVO: Analizando {cantidad} portafolios."

    print(f"\n[{estado_ia}]")
    print("1. Registrar nueva inversión")
    print("2. Listar inversiones (Tabla)")
    print("3. Analizar portafolios con IA")
    print("4. Borrar base de datos")
    print("5. Apagar sistema")

    opcion = input("\nInstrucción de usuario: ")

    if opcion == "1":
        print("\n>>> CAPTURA DE DATOS (Interés Compuesto):")
        nombre = input("Nombre del cliente: ")
        cap = input("Capital inicial ($): ")
        tas = input("Tasa de interés anual (%): ")
        tie = input("Tiempo de inversión (años): ")

        # Instanciación: Creación del objeto [4, 16]
        nueva_inversion = Interes_compuesto(nombre, cap, tas, tie)
        nueva_inversion.calcular_monto() # Llama a la función de cálculo
        lista_inversiones.append(nueva_inversion)
        print(f">>> IA: Inversión de '{nombre}' procesada con éxito.")

    elif opcion == "2":
        if len(lista_inversiones) > 0:
            # Encabezado de tabla alineado [3, 17]
            print("\n" + "="*85)
            print(f"{'CLIENTE':<15} | {'CAPITAL':<10} | {'TASA %':<8} | {'AÑOS':<8} | {'MONTO FINAL':<12} | {'GANANCIA':<10}")
            print("-" * 85)

            # Ciclo for para recorrer objetos [18, 19]
            for inv in lista_inversiones:
                print(inv.obtener_fila_tabla())
            print("="*85)
        else:
            print("\n>>> IA: No existen registros para generar la tabla.")

    elif opcion == "3":
        if len(lista_inversiones) > 0:
            print("\n>>> IA REALIZANDO ANÁLISIS ESTRATÉGICO:")
            for inv in lista_inversiones:
                inv.mostrar_informe()
        else:
            print("\n>>> IA: Memoria vacía. Nada que analizar.")

    elif opcion == "4":
        lista_inversiones = []
        print("\n>>> IA: Registros eliminados. Memoria en cero.")

    elif opcion == "5":
        print("\n>>> IA: Desconectando protocolos... Adiós.")
        break # Rompe el ciclo para salir [20]

    else:
        print("\n>>> IA ERROR: Comando no válido.")

print("\n--- SISTEMA FUERA DE LÍNEA ---")
```
